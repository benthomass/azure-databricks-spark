```sql
%python
df = spark.read.parquet(f"{presentation_folder_path}/race_results")

df.write.mode("overwrite") \
  .format("parquet") \
  .option("path", f"{presentation_folder_path}/race_results_ext_python") \
  .saveAsTable("objectsdemo.race_results_ext_python")
```
